# Tokenizer

[Huggingface](https://huggingface.co/) provides a tokenizer module that we can use to train our own one. In the next notebook, we will be training a model to translate English into Chinese, therefore, we will first train two tokenizers, one on each language.

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

en_tokenizer = Tokenizer(BPE())
en_tokenizer.pre_tokenizer = Whitespace()
trainer = BpeTrainer(special_tokens=["[UNK]", "[CLS]", "[SEP]", "[PAD]", "[MASK]"])
en_tokenizer.train(files=["./data/en-zh.txt/TED2013.en-zh.en"], trainer=trainer)
en_tokenizer.save("./output/ted-english.json")
```

```python
zh_tokenizer = Tokenizer(BPE())
trainer = BpeTrainer(special_tokens=["[UNK]", "[CLS]", "[SEP]", "[PAD]", "[MASK]"])
zh_tokenizer.train(files=["./data/en-zh.txt/TED2013.en-zh.zh"], trainer=trainer)
zh_tokenizer.save("./output/ted-chinese.json")
```

After saving it, we can always load and use it in the future

```python
en_tokenizer = Tokenizer.from_file("./output/ted-english.json")
output = en_tokenizer.encode("That's the theme of my talk.")
print(output.tokens)
```

```python
zh_tokenizer = Tokenizer.from_file("./output/ted-chinese.json")
output = zh_tokenizer.encode("这就是我演讲的主题。")
print(output.tokens)
```
